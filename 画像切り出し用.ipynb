# 画像分割

```python
import os
from PIL import Image

# 対象フォルダ
folder_path = "datasets/Radish/ProcessingPlace"

# 対応拡張子
valid_exts = (".png", ".jpg", ".jpeg", ".webp")

for filename in os.listdir(folder_path):
    if filename.lower().endswith(valid_exts):
        file_path = os.path.join(folder_path, filename)

        # 画像読み込み
        img = Image.open(file_path)
        img = img.convert("RGB")

        width, height = img.size

        # 各領域の割合指定(面積1/3くらい被っててOK)（ex.端から60%点で分割）
        # 書式　"名前末尾": (x1, y1, x2, y2) ただしx1<x2
        
        regions = {
            "A": (0.0, 0.0, 0.6, 0.6),
            "B": (0.0, 0.4, 0.6, 1.0),
            "C": (0.4, 0.0, 1.0, 0.6),            
            "D": (0.4, 0.4, 1.0, 1.0),
           
        }
        """
        
        regions = {
            "A": (0.0, 0.0, 0.43, 0.43),
            "B": (0.0, 0.29, 0.43, 0.71),
            "C": (0.0, 0.57, 0.43, 1.0),            
            "D": (0.29, 0.0, 0.71, 0.43),
            "E": (0.29, 0.29, 0.71, 0.71),
            "F": (0.29, 0.57, 0.71, 1.0),            
            "G": (0.57, 0.0, 1.0, 0.43),
            "H": (0.57, 0.29, 1.0, 0.71),
            "I": (0.57, 0.57, 1.0, 1.0),   
           
        }
        """


        

        name, _ = os.path.splitext(filename)

        for suffix, (lx, uy, rx, ly) in regions.items():
            left   = int(width  * lx)
            upper  = int(height * uy)
            right  = int(width  * rx)
            lower  = int(height * ly)

            cropped_img = img.crop((left, upper, right, lower))

            save_name = f"{name}_{suffix}.png"
            save_path = os.path.join(folder_path, save_name)

            cropped_img.save(save_path, format="PNG")

            print(f"saved: {save_name}")
```

# ファイル名変更

```python
import os
import re

# ===== 設定 =====
folder_path = "datasets/Radish/ProcessingPlace"
start_number = 331

# ===== 対象ファイル抽出 =====
pattern = re.compile(r"^Radish_Hatsukadaikon_.*_.*\.png$")

files = [
    f for f in os.listdir(folder_path)
    if pattern.match(f)
]

# 並び順を安定させる（重要）
files.sort()

print("対象ファイル数:", len(files))

# ===== リネーム処理 =====
current_number = start_number

for filename in files:
    old_path = os.path.join(folder_path, filename)
    new_filename = f"Radish_Hatsukadaikon_{current_number}.png"
    new_path = os.path.join(folder_path, new_filename)

    # 既存ファイルと衝突しないよう確認
    if os.path.exists(new_path):
        print(f"⚠ 既に存在: {new_filename} → スキップ")
        continue

    os.rename(old_path, new_path)
    print(f"{filename} → {new_filename}")

    current_number += 1

print("完了")
```
